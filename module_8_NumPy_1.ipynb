ndarray

```python
import numpy as np

# 1d array
arr1 = np.array([1,2,3,4,5])
print(type(arr1))
print(arr1.ndim)

# 2d array
arr2 = np.array([[1,2,3],  # row1
                 [4,5,6]]) # row2
print(arr2.ndim)

# 3d array
arr3 = np.array([
    [
        [1,2,3],
        [4,5,6]
    ],
    [
        [7,2,9],
        [8,5,3]
    ],
    [
        [5,6,3],
        [1,2,6]
    ]
])
print(arr3.ndim)
```

Array Attributes

```python
# dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

# shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

# data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

# size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

ndarray data type

```python
arr = np.array([1,2,3])
print(arr.dtype)

arr = np.array([1.3,45.4,2.4])
print(arr.dtype)
```

```python
arr = np.array([1,2,3.4])
# upcasting
print(arr.dtype)

# upcasted to string
arr = np.array([1,2,3.4,"ss"])
print(arr.dtype)
```

Selecting a data type for an array

```python
arr = np.array([1,2,300])
print(arr.dtype)
arr = arr.astype(np.int16)
print(arr.dtype)

arr = np.array([1,2,True,300,"hello"])
print(arr.dtype)
```

ndarray creation from existing data

```python
# list
list1 = [10,20,30,40.5]
arr = np.array(list1, np.int32)
print(arr)
print(type(arr))
print(arr.dtype)

mixed_list = [10,True,'Hello',10.5]
arr = np.array(mixed_list)
print(arr)
print(type(arr))
print(arr.dtype)

matrix = [
    [1,2,3],
    [4,5,7]
]
arr = np.array(matrix)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# tuple
tpl = (1,2,3)
arr = np.array(tpl)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# set
set1 = {1,2,3}
arr = np.array(list(set1))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# disctionary
dict1 = {"a":1, "b":2, "c":3}
print(type(dict1))
keys = dict1.keys()
values = dict1.values()
items = dict1.items()
print(keys, values, items)
```

```python
# keys to ndarray
arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# values to ndarray
arr = np.array(list(values))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# items to ndarray
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
