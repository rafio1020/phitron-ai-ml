Mathematical Functions

```python
# trigonometry
#sin cos tan takes the value in radian
sin_val = np.sin(x)
print(sin_val)

cos_val = np.cos(x)
print(cos_val)

deg_con = np.rad2deg(x) # radian to degree
print(deg_con)

#deg2rad
```

```python
# log
x = np.array([1,10,100,1000])
base_10_log_val = np.log10(x)
print(base_10_log_val)

# log base 2
base_2_log_val = np.log2(x)
print(base_2_log_val)
```

```python
# different
sum = np.sum(x)
cumulative_sum = np.cumsum(x)
print(sum,cumulative_sum)
```

Logical Function


```python
# Comparison
x = np.array([10,8,16,100])
y = np.array([2,10,5,6])
greater_than = x>y
print(greater_than)

# like this determine others like <,==,<=,>=,!=
# all() --> Checks if all are true
# any() --> checks if one if them are true

print(np.all(x>y))
print(np.any(x>y))
```

Sorting

```python
#inplace
x = np.array([10,8,16,100])
z = x.copy()
print(z)
z.sort()
print(z)

#copy sorting
sort_arr = np.sort(x)
print(sort_arr)
```

```python
# 2nd array sorting
mat = np.array([
    [20,30,30],
    [10,5,60]
])

# horizon axis = 1
hor_sort = np.sort(mat,axis=1)
print(hor_sort)

# vertical axis = 0
ver_sort = np.sort(mat,axis=0)
print(ver_sort)
```

Searching

```python
print(x)
```

```python
# np.where(condition) ; returns index

index = np.where(x==16)
print(index)

index = np.where(x>8)
print(index)

# np.where(condition,x,y) if true x, otherwise replace with y; returns the array not the index
arr = np.where(x>8,x,0)
print(arr)
```

Statistical Analysis

```python
# dataset load
data = np.genfromtxt('sample_data/student_scores.csv', delimiter=',', skip_header=1)
print(data)
```

```python
math_marks = data[::,:1]
print(math_marks.T)

max_math_marks = np.max(math_marks)
print(max_math_marks)

min_math_marks = np.min(math_marks)
print(min_math_marks)

average_math_marks = np.mean(math_marks)
print(average_math_marks)

median_math_marks = np.median(math_marks)
print(median_math_marks)

# standard deviation
std_math_marks = np.std(math_marks)
print(std_math_marks)
```

```python
# Correlation
study_hours = np.array([2,4,5,6])
exam_scores = np.array([40,50,60,80])

data = np.array([study_hours,exam_scores])
print(data)

correlation = np.corrcoef(data)
print(correlation)
```

Linear Algebra


```python
# matrics multiplication

A = np.array([
    [4,5,6],
    [7,8,9],
    [1,2,3]
])

B = np.array([
    [1,2,3],
    [14,9,6],
    [7,8,9]
])

dot_product = np.dot(A,B)
print(dot_product)

# trace
print(np.trace(B)) # gets the sum of the diagonal elements

# determinant
print(np.linalg.det(B))

# rank
rank_of_B = np.linalg.matrix_rank(B)
print(rank_of_B)
```
